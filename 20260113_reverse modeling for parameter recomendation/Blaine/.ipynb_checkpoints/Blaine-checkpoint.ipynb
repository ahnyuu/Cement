```python
directory = 'database/'
df = pd.read_csv(directory + 'normalized_cmall_remove_duplicate_m_all.csv') #df ; data flame, pd ; pandas

# INPUT DATA
#file_name coulumn을 선택
feed_total = df['feed_total'].values
feed_clinker = df['feed_clinker'].values
feed_gypsum = df['feed_gypsum'].values
feed_slag = df['feed_slag'].values
feed_FA = df['feed_FA'].values
RP_spac = df['RP_spac'].values
RP_skew = df['RP_skew'].values
RP_roller_p1 = df['RP_roller_p1'].values
RP_roller_p2 = df['RP_roller_p2'].values
RP_roller_energy1 = df['RP_roller_energy1'].values
RP_roller_energy2 = df['RP_roller_energy2'].values
RP_roller_vib = df['RP_roller_vib'].values
RP_BE_energy1 = df['RP_BE_energy1'].values
RP_BE_energy2 = df['RP_BE_energy2'].values
RP_sep_rpm = df['RP_sep_rpm'].values
RP_sep_fan_damper = df['RP_sep_fan_damper'].values
RP_sep_BF_pressure = df['RP_sep_BF_pressure'].values
RP_sep_BF_damper = df['RP_sep_BF_damper'].values
dosing_BE_energy = df['dosing_BE_energy'].values
mill_feed_c = df['mill_feed_c'].values
mill_feed_cir = df['mill_feed_cir'].values
mill_energy = df['mill_energy'].values
mill_in_temp = df['mill_in_temp'].values
mill_out_temp = df['mill_out_temp'].values
mill_BE_energy = df['mill_BE_energy'].values
mill_out_gas_temp = df['mill_out_gas_temp'].values
mill_out_mater_temp = df['mill_out_mater_temp'].values
mill_BF_pressure = df['mill_BF_pressure'].values
mill_BF_damper = df['mill_BF_damper'].values
mill_sep_rpm = df['mill_sep_rpm'].values
mill_sep_fan_damper = df['mill_sep_fan_damper'].values
mill_sep_BF_pressure = df['mill_sep_BF_pressure'].values
mill_sep_BF_fan_damper = df['mill_sep_BF_fan_damper'].values
final_BE_1 = df['final_BE_1'].values
final_BE_2 = df['final_BE_2'].values
final_mater_temp = df['final_mater_temp'].values
grind_aid = df['grind_aid'].values
blaine_prev = df['blaine_prev'].values
residue_prev = df['residue_prev'].values


# OUTPUT DATA
blaine = df['blaine'].values
residue = df['residue'].values

# INPUT DATA
print(feed_clinker.shape)

# OUTPUT DATA
print(blaine.shape)
print(residue.shape)
```

```python
feed_total = np.array(feed_total)
feed_clinker = np.array(feed_clinker)
feed_gypsum = np.array(feed_gypsum)
feed_slag = np.array(feed_slag)
feed_FA = np.array(feed_FA)
RP_spac = np.array(RP_spac)
RP_skew = np.array(RP_skew)
RP_roller_p1 = np.array(RP_roller_p1)
RP_roller_p2 = np.array(RP_roller_p2)
RP_roller_energy1 = np.array(RP_roller_energy1)
RP_roller_energy2 = np.array(RP_roller_energy2)
RP_roller_vib = np.array(RP_roller_vib)
RP_BE_energy1 = np.array(RP_BE_energy1)
RP_BE_energy2 = np.array(RP_BE_energy2)
RP_sep_rpm = np.array(RP_sep_rpm)
RP_sep_fan_damper = np.array(RP_sep_fan_damper)
RP_sep_BF_pressure = np.array(RP_sep_BF_pressure)
RP_sep_BF_damper = np.array(RP_sep_BF_damper)
dosing_BE_energy = np.array(dosing_BE_energy)
mill_feed_c = np.array(mill_feed_c)
mill_feed_cir = np.array(mill_feed_cir)
mill_energy = np.array(mill_energy)
mill_in_temp = np.array(mill_in_temp)
mill_out_temp = np.array(mill_out_temp)
mill_BE_energy = np.array(mill_BE_energy)
mill_out_gas_temp = np.array(mill_out_gas_temp)
mill_out_mater_temp = np.array(mill_out_mater_temp)
mill_BF_pressure = np.array(mill_BF_pressure)
mill_BF_damper = np.array(mill_BF_damper)
mill_sep_rpm = np.array(mill_sep_rpm)
mill_sep_fan_damper = np.array(mill_sep_fan_damper)
mill_sep_BF_pressure = np.array(mill_sep_BF_pressure)
mill_sep_BF_fan_damper = np.array(mill_sep_BF_fan_damper)
final_BE_1 = np.array(final_BE_1)
final_BE_2 = np.array(final_BE_2)
final_mater_temp = np.array(final_mater_temp)
grind_aid = np.array(grind_aid)
blaine_prev = np.array(blaine_prev)
residue_prev = np.array(residue_prev)

blaine = np.array(blaine)
residue = np.array(residue)
```

```python
data_x = tf.stack([feed_total, feed_clinker, feed_gypsum, feed_slag, feed_FA, RP_spac, RP_skew,
    RP_roller_p1, RP_roller_p2, RP_roller_energy1, RP_roller_energy2, RP_roller_vib, RP_BE_energy1, RP_BE_energy2,
    RP_sep_rpm, RP_sep_fan_damper, RP_sep_BF_pressure, RP_sep_BF_damper, dosing_BE_energy, mill_feed_c, mill_feed_cir,
    mill_energy, mill_in_temp, mill_out_temp, mill_BE_energy, mill_out_gas_temp, mill_out_mater_temp, mill_BF_pressure,
    mill_BF_damper, mill_sep_rpm, mill_sep_fan_damper, mill_sep_BF_pressure, mill_sep_BF_fan_damper, final_BE_1, final_BE_2, 
    final_mater_temp, grind_aid, blaine_prev, residue_prev], 1)

data_y = tf.stack([blaine, residue], 1)

print(data_x.shape)
print(data_y.shape)
```

```python
n_data = data_x.shape[0]

data_x = np.reshape(data_x, [n_data,-1])
data_y = np.reshape(data_y, [n_data,-1])
```

```python
data_x = np.asmatrix(data_x)
data_y = np.asmatrix(data_y)

print(data_x.shape)
print(data_y.shape)
print(data_x)
```

```python
# 데이터셋 나누기 (70% 학습, 15% 검증, 15% 테스트)
train_x, temp_x, train_y, temp_y = train_test_split(data_x, data_y, test_size=0.3, random_state=42)
val_x, test_x, val_y, test_y = train_test_split(temp_x, temp_y, test_size=0.5, random_state=42)
```

# Parametric Study
    # number of layers : 7
    # node shapes : 128, 1024, 2048, 2048, 1024, 64, 8
    # mini-batch size : 512
    # dropout shapes : 0.1,  0.2,0.3, 0.2, 0.2, 0.2, 0.0
    # actication func : swish

```python
# 모델 정의
def build_model(input_shape):
    model = Sequential([
        Dense(128, activation='swish', kernel_initializer='glorot_uniform', input_shape=input_shape),
        BatchNormalization(),
        Dropout(0.1),  # 20% Dropout

        Dense(1024, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(2048, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.3),  # 깊은 레이어에서는 30%로 증가

        Dense(2048, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(1024, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(64, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(8, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.0),  # 얕은 레이어에서는 Dropout을 낮게

        Dense(2)  # 출력 레이어
    ])
    return model

# 모델 생성 및 컴파일
input_shape = (39,)
model = build_model(input_shape)
huber_loss = tf.keras.losses.Huber(delta=1.0)

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005), loss=huber_loss)

# ModelCheckpoint 콜백: 최적의 Epoch 모델 저장
checkpoint = tf.keras.callbacks.ModelCheckpoint(
    'best_model.h5',        # 최적 모델 저장 경로                              # 저장 파일 이름 설정
    monitor='val_loss',     # 검증 손실 기준
    save_best_only=True,    # 가장 낮은 val_loss 모델만 저장
    save_weights_only=False, # 모델 전체 저장 (구조+가중치+옵티마이저 상태)
    mode='min',             # val_loss가 작을수록 좋음
    verbose=1
)

# 조기 종료 콜백
early_stopping = EarlyStopping(monitor='val_loss', patience=300, verbose=1)
```

```python
# 모델 요약 출력
model.summary()
```

```python
# 모델 학습
history = model.fit(
    train_x, train_y,
    epochs=1000,
    batch_size=512,
    validation_data=(val_x, val_y),
    callbacks=[early_stopping, checkpoint]  # 콜백에 ModelCheckpoint 추가
)

# 최적의 모델 불러오기
best_model = tf.keras.models.load_model('best_model.h5')                     # 저장 파일 이름 설정

# 모델 평가
test_loss = best_model.evaluate(test_x, test_y)
print(f'Test Loss (Best Model): {test_loss}')

# 학습 결과 시각화
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
#plt.ylim(0.0, 0.002)   #y축 범위 설정
plt.legend()
plt.show()
```

# Confusion matrix 계산 (Blaine and Residue total)

```python
# 실제값 (Ground Truth)과 예측값 (Predicted Values) 변환
actual_blaine = test_y[:, 0]
actual_blaine = (((actual_blaine - 0.1) / 0.8) * 1333) + 3127  # 실제 Blaine 값
predicted_blaine = best_model.predict(test_x)[:, 0]
predicted_blaine = (((predicted_blaine - 0.1) / 0.8) * 1333) + 3127  # 예측된 Blaine 값

actual_residue = test_y[:, 1]
actual_residue = (((actual_residue - 0.1) / 0.8) * 7.6) + 3.7  # 실제 Residue 값
predicted_residue = best_model.predict(test_x)[:, 1]
predicted_residue = (((predicted_residue - 0.1) / 0.8) * 7.6) + 3.7  # 예측된 Residue 값

# Blaine 및 Residue를 OK/NG로 변환하는 함수
def categorize_blaine_residue(blaine_values, residue_values):
    categories = []
    for blaine, residue in zip(blaine_values, residue_values):
        blaine_status = "OK" if 3700 <= blaine <= 3900 else "NG"
        residue_status = "OK" if 7 <= residue <= 9 else "NG"
        categories.append(f"{blaine_status}-{residue_status}")
    return np.array(categories)

# 변환된 값 (실제값과 예측값을 OK/NG 조합으로 변환)
actual_labels = categorize_blaine_residue(actual_blaine, actual_residue)
predicted_labels = categorize_blaine_residue(predicted_blaine, predicted_residue)

# Confusion Matrix 및 Accuracy 계산
labels = ["OK-OK", "OK-NG", "NG-OK", "NG-NG"]
conf_matrix = confusion_matrix(actual_labels, predicted_labels, labels=labels)
accuracy = accuracy_score(actual_labels, predicted_labels)

# 결과 출력
df_conf_matrix = pd.DataFrame(conf_matrix, index=[f"Actual {lbl}" for lbl in labels], 
                              columns=[f"Predicted {lbl}" for lbl in labels])

print("Confusion Matrix:")
print(df_conf_matrix)
print(f"\nAccuracy: {accuracy:.4f}")
```

```python
# 모델 정의
def build_model(input_shape):
    model = Sequential([
        Dense(128, activation='swish', kernel_initializer='glorot_uniform', input_shape=input_shape),
        BatchNormalization(),
        Dropout(0.1),  # 20% Dropout

        Dense(1024, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(2048, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.3),  # 깊은 레이어에서는 30%로 증가

        Dense(2048, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(1024, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(64, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.2),

        Dense(8, activation='swish', kernel_initializer='glorot_uniform'),
        BatchNormalization(),
        Dropout(0.0),  # 얕은 레이어에서는 Dropout을 낮게

        Dense(2)  # 출력 레이어
    ])
    return model

# 모델 생성 및 컴파일
input_shape = (39,)
model = build_model(input_shape)
huber_loss = tf.keras.losses.Huber(delta=1.0)

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005), loss=huber_loss)

# ModelCheckpoint 콜백: 최적의 Epoch 모델 저장
checkpoint = tf.keras.callbacks.ModelCheckpoint(
    'best_model_bs1024.h5',        # 최적 모델 저장 경로
    monitor='val_loss',     # 검증 손실 기준
    save_best_only=True,    # 가장 낮은 val_loss 모델만 저장
    save_weights_only=False, # 모델 전체 저장 (구조+가중치+옵티마이저 상태)
    mode='min',             # val_loss가 작을수록 좋음
    verbose=1
)

# 조기 종료 콜백
early_stopping = EarlyStopping(monitor='val_loss', patience=300, verbose=1)

# 모델 학습
history = model.fit(
    train_x, train_y,
    epochs=1000,
    batch_size=1024,
    validation_data=(val_x, val_y),
    callbacks=[early_stopping, checkpoint]  # 콜백에 ModelCheckpoint 추가
)

# 최적의 모델 불러오기
best_model = tf.keras.models.load_model('best_model_bs1024.h5')

# 모델 평가
test_loss = best_model.evaluate(test_x, test_y)
print(f'Test Loss (Best Model): {test_loss}')

# 학습 결과 시각화
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
#plt.ylim(0.0, 0.002)   #y축 범위 설정
plt.legend()
plt.show()
```
